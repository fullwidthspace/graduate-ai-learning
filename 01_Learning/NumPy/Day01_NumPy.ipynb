```python
import pandas as pd

data = {
    "ship": ["Ship_A", "Ship_B", "Ship_C"],
    "speed": [12.5, 14.2, 11.8],
    "fuel": [35.2, 42.1, 31.5]
}

df = pd.DataFrame(data)

df
```

```python
import matplotlib.pyplot as plt

plt.plot(df["speed"], df["fuel"], marker="o")

plt.xlabel("Speed")
plt.ylabel("Fuel Consumption")
plt.title("Ship Speed vs Fuel Consumption")

plt.show()
```
